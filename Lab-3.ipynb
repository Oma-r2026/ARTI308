```python
plt.figure(figsize=(8,5))
sns.histplot(df['price'], bins=20)
plt.title("Distribution of Prices")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['mileage'], bins=20)
plt.title("Distribution of Mileages")
plt.show()
```

```python
#I changed the sum() function to mean() because sum will just add up however many times a car appears, to adjust 
# per model i used mean. I also changed the plot name to model_avg_price
model_avg_price = df.groupby('model')['price'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))
model_avg_price.plot(kind='bar')
plt.title("Average Prices for models")
plt.ylabel("avg Prices")
plt.show()

model_avg_price
```

```python
#changed stuff again to fit my dataset.
# This displays the average mpg for a car model to see which model is most fuel efficent 
model_mpg = df.groupby('model')['mpg'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))
model_mpg.plot(kind='bar')
plt.title("Average MPG by Model")
plt.ylabel("Average MPG")
plt.show()

model_mpg
```

```python
# plt.figure(figsize=(8,5))
# sns.scatterplot(x='Boxes Shipped', y='Amount', data=df)
# plt.title("Boxes Shipped vs Revenue")
# plt.show()
# Converted this to fit my dataset. Assuming negative correlation with higher mileage, prices lowers
plt.figure(figsize=(8,5))
sns.scatterplot(x='mileage', y='price', data=df)
plt.title("Mileage vs Price")
plt.show()
```

```python
# plt.figure(figsize=(6,4))
# sns.heatmap(df[['Boxes Shipped', 'Amount']].corr(), annot=True)
# plt.title("Correlation Matrix")
# plt.show()
# Again converted to fit my dataset

plt.figure(figsize=(6,4))
sns.heatmap(df[['mileage', 'price']].corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
year_price = df.groupby('year')['price'].mean().sort_index()

plt.figure(figsize=(10,5))
year_price.plot()
plt.title("Average Price by Car Year")
plt.ylabel("Average Price")
plt.show()
```
